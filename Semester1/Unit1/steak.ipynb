### My Question
Does gambling increase a persons chance of cheating on their significant other?

### My Study
I will use `for` loops to do an observational study. This will not be based on gender but just in general.

### My Analysis

```python
import seaborn as sns
import pandas as pd

sns.set_theme('paper')
sns.set_style('darkgrid')

df = pd.read_csv(r'C:\Users\*******\!jupyter\data\steak.csv')

cheats_and_gambles = 0
loyal_and_gambles = 0

for i in range(len(df)):
    if str(df['Do you ever gamble?'][i]).lower() == "yes" and str(df['Have you ever cheated on your significant other?'][i]).lower() == "yes":
        cheats_and_gambles += 1
    elif str(df['Do you ever gamble?'][i]).lower() == "yes" and str(df['Have you ever cheated on your significant other?'][i]).lower() == "no":
        loyal_and_gambles += 1
    else:
        continue

print(f"cheats: {cheats_and_gambles} | loyal: {loyal_and_gambles}")

sns.countplot(
    y=["Cheats & Gambles"] * cheats_and_gambles +
      ["Loyal & Gambles"] * loyal_and_gambles,
      color="#ff9a2a"
)
```

### My Conclusion
In conclusion, gamblers who have not cheated were more common than gamblers who have cheated. The frequency of loyal gamblers was 30 compared to 20 cheaters, and their relative frequencies were 60% and 40%, respectively. Therefore, the data suggests that gamblers in this data were more likely to report being loyal than having cheated.
