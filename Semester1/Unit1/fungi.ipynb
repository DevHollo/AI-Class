# Question:
How many mushrooms are edible vs poisonus in the meadows habitat

```python
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\*******\!jupyter\data\mushrooms.csv')
```

```python
sns.set_theme('paper')
sns.set_style('darkgrid')

poision = 0
edible = 0

for i in range(len(df)):
    if str(df.habitat[i]).lower() == "m" and str(df.Class[i]).lower() == "e":
        edible += 1
    elif str(df.habitat[i]).lower() == "m" and str(df.Class[i]).lower() == "p":
        poision += 1
    else:
        continue

counts = df[df["habitat"].str.lower() == "m"]["Class"].value_counts().reset_index() # ChatGPT helped me get an idea on how to do this variable

sns.barplot(data=counts, x="count", y="Class", color="#ff9a2a")
print(f"p: {poision} | e: {edible}")
```

## Conclusion:
In the meadows habitat there are 256 edible mushrooms and 36 poision mushrooms. This means there are more edible than poision.

Frequency: `292`
Edible relative frequency: `87.67%`
Poision relative frequency: `12.32%`
